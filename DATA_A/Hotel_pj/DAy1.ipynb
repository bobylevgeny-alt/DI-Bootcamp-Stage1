```python
occupancy_analysis = df[[
    'Month',
    'Sold occupancy',
    'Available occupancy',
    'Occupancy rate (%)',
    'Average room occupancy',
    'Average length of stay',
    'N. unique guests'
]]
occupancy_analysis
```

Occupancy analysis reveals a pronounced seasonality in the hotel's operations. The highest occupancy rates were observed in August 2026 (53.11%) and December 2025 (46.65%). In most other months, occupancy rates ranged between 30% and 40%. The average length of stay is approximately 1–1.5 nights, indicating a predominance of short-term trips. The average number of guests per occupied room is approximately two. March and part of April 2026 should be considered separately, as the figures for this period were significantly distorted by the unstable external environment.

Less dependence on OTAs, more clear direct channels, better source capture for manual bookings, dynamic price management by season, and separate cancellation controls.
