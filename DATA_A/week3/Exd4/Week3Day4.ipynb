```python
import matplotlib.pyplot as plt
```

```python
dayofweek = [1,2,3,4,5,6,7]
temperature = [22,23,24,25,26,27,28]
plt.plot(dayofweek, temperature, color='red')
plt.title('Temperature in this week')
plt.ylabel('Temperature')
plt.show()
```

```python
population = [100,35,64,43,77,54]
cities = ['Tel Aviv', 'Haifa', 'Holon', 'Moscow', 'Hadera', 'Akko']

plt.bar(cities, population)
plt.show()
```

```python
grades  = np.random.randint(low=0, high=101,size = 50)
print(grades)
```

```python
ice_cream_flavor = ['Caramel', 'Choco', 'Pist', 'Vanil']
people = [4,5,7,2]
explode = (0,0.5,0,0)
plt.pie(people, labels=ice_cream_flavor, autopct='%1.1f%%', explode=explode)
plt.show()
```
